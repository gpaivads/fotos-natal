```python
#TRATAR LINKS

import re

diretorio = 'natal26'

# ===== CONFIGURAÇÕES =====
arquivos_origem = [
    f"{diretorio}/links.txt"
]

arquivo_saida = f"{diretorio}/links_tratados.txt"
width_novo = 1920
height_novo = 1080
# =========================

# Regex: pega QUALQUER URL do domínio images-proof.alboompro.com
regex_url = re.compile(
    #r'https://images-proof\.alboompro\.com/[^\s"<>]+'
    r'https://cdn-selpics-b2.youfocus.com.br/[^\s"<>]+'
)

links_final = []

for arquivo in arquivos_origem:
    with open(arquivo, "r", encoding="utf-8", errors="ignore") as f:
        conteudo = f.read()

        encontrados = regex_url.findall(conteudo)

        for url in encontrados:
            # Ajusta width e height, independentemente do valor atual
            url = re.sub(r'width/\d+', f'width/{width_novo}', url)
            url = re.sub(r'height/\d+', f'height/{height_novo}', url)

            links_final.append(url)

# Remove duplicados preservando ordem
links_final = list(dict.fromkeys(links_final))

with open(arquivo_saida, "w", encoding="utf-8") as f:
    for link in links_final:
        f.write(link + "\n")

print(f"✅ {len(links_final)} links extraídos e salvos em '{arquivo_saida}'")
```

```python
# BAIXAR IMAGENS 2025

import os
import requests

# ===== CONFIGURAÇÕES =====
arquivo_links = arquivo_saida
pasta_destino = f"{diretorio}/imagens"
prefixo_nome = diretorio
timeout_segundos = 30
# =========================

os.makedirs(pasta_destino, exist_ok=True)

with open(arquivo_links, "r", encoding="utf-8") as f:
    links = [linha.strip() for linha in f if linha.strip()]

print(f"Total de links encontrados: {len(links)}")

for idx, url in enumerate(links, start=1):
    nome_arquivo = f"{prefixo_nome}_{idx}.jpg"
    caminho_arquivo = os.path.join(pasta_destino, nome_arquivo)

    try:
        print(f"Baixando {nome_arquivo}...")
        resposta = requests.get(url, timeout=timeout_segundos)
        resposta.raise_for_status()

        with open(caminho_arquivo, "wb") as img:
            img.write(resposta.content)

    except Exception as e:
        print(f"❌ Erro ao baixar {url}")
        print(e)

print("✅ Download finalizado!")
```
