```python
import os
import requests

# Nome do arquivo que contém as URLs
input_file = "links.txt"

# Diretório para salvar as imagens baixadas
output_directory = "imagens"
os.makedirs(output_directory, exist_ok=True)

# Função para baixar uma imagem
def download_image(url, output_dir):
    try:
        # Nome do arquivo baseado na URL
        file_name = os.path.join(output_dir, url.split("/")[-1])
        
        # Requisição para download
        response = requests.get(url, stream=True, timeout=10)
        response.raise_for_status()
        
        # Salvar o conteúdo em um arquivo
        with open(file_name, "wb") as file:
            for chunk in response.iter_content(chunk_size=8192):
                file.write(chunk)
        print(f"Imagem salva: {file_name}")
    except requests.RequestException as e:
        print(f"Erro ao baixar {url}: {e}")

# Ler as URLs do arquivo e fazer o download
with open(input_file, "r", encoding="utf-8") as file:
    urls = [line.strip() for line in file if line.strip()]  # Remove linhas em branco

print(f"{len(urls)} URLs encontradas. Iniciando downloads...")

for url in urls:
    download_image(url, output_directory)

print("Downloads concluídos.")
```

```python
import os

# Diretório onde estão os arquivos .jpg
input_directory = "imagens"  # Substitua pelo caminho correto

# Função para renomear arquivos
def rename_files(directory):
    try:
        for file_name in os.listdir(directory):
            # Verificar se o arquivo é um .jpg
            if file_name.endswith(".JPG"):
                # Encontrar o índice de "_DSC"
                index = file_name.find("DSC")
                if index != -1:
                    # Novo nome removendo tudo antes de "_DSC"
                    base_name = file_name[index:]
                    new_name = base_name
                    old_path = os.path.join(directory, file_name)
                    new_path = os.path.join(directory, new_name)
                    
                    # Resolver duplicidades
                    counter = 1
                    while os.path.exists(new_path):
                        name, ext = os.path.splitext(base_name)
                        new_name = f"{name}_{counter}{ext}"
                        new_path = os.path.join(directory, new_name)
                        counter += 1
                    
                    # Renomear o arquivo
                    os.rename(old_path, new_path)
                    print(f"Renomeado: {file_name} -> {new_name}")
                else:
                    print(f"Arquivo ignorado (não contém '_DSC'): {file_name}")
    except Exception as e:
        print(f"Erro ao renomear arquivos: {e}")

# Chamar a função
rename_files(input_directory)
```

```python
import os
import requests
import time

# Arquivo de entrada com as URLs
input_file = "links2.txt"

# Arquivo de saída para salvar URLs únicas
output_file = "urls_unicas.txt"

# Diretório para salvar as imagens baixadas
output_directory = "imagens"
os.makedirs(output_directory, exist_ok=True)

# Função para processar URLs e lidar com duplicidades
def process_urls(input_file, output_file, output_directory):
    unique_urls = {}  # Dicionário para armazenar URLs únicas com base no trecho "_DSC"
    downloaded = 0

    # Ler URLs do arquivo
    with open(input_file, "r", encoding="utf-8") as file:
        urls = [line.strip() for line in file if line.strip()]

    # Processar cada URL
    for url in urls:
        # Encontrar o índice de "_DSC"
        index = url.find("DSC")
        if index != -1:
            identifier = url[index:]  # Capturar tudo a partir de "_DSC"
            if identifier not in unique_urls:
                unique_urls[identifier] = url  # Adicionar a primeira URL encontrada

    # Salvar URLs únicas em um novo arquivo
    with open(output_file, "w", encoding="utf-8") as file:
        for url in unique_urls.values():
            file.write(url + "\n")

    print(f"URLs únicas salvas em '{output_file}'. Iniciando downloads...")

    # Fazer download das imagens
    for identifier, url in unique_urls.items():
        try:
            response = requests.get(url, stream=True, timeout=10)
            response.raise_for_status()

            # Salvar o arquivo com base no identificador
            file_name = os.path.join(output_directory, identifier)
            with open(file_name, "wb") as img_file:
                for chunk in response.iter_content(chunk_size=8192):
                    img_file.write(chunk)

            print(f"Imagem baixada: {file_name}")
            downloaded += 1
        except requests.RequestException as e:
            print(f"Erro ao baixar {url}: {e}")

        # Pausa de 1 segundo entre os downloads
        time.sleep(1)

    print(f"Downloads concluídos. Total de imagens baixadas: {downloaded}")

# Executar o script
process_urls(input_file, output_file, output_directory)
```
